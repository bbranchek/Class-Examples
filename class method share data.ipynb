```python
class ProductCatalog:
    _products = {}   # shared class-level dictionary

    @classmethod
    def load_products(cls, product_list):
        for item in product_list:
            cls._products[item["sku"]] = item

    @classmethod
    def add_product(cls, sku, name, price):
        cls._products[sku] = {"sku": sku, "name": name, "price": price}

    @classmethod
    def all_products(cls):
        return list(cls._products.values())
```

```python
# Load initial shared data
# ------------------------------
initial_data = [
    {"sku": "A100", "name": "USB Cable", "price": 5.99},
    {"sku": "B200", "name": "Keyboard", "price": 29.99},
]

ProductCatalog.load_products(initial_data)
```

```python
# ------------------------------
# Create an instance
# ------------------------------
pc = ProductCatalog()

# Print already-loaded shared data
print("Initial catalog:")
print(pc.all_products())
```

```python

# ------------------------------
# Add new data
# ------------------------------
pc.add_product("C300", "Wireless Mouse", 19.99)
```

```python
# Print everything again
print("\nUpdated catalog:")
print(pc.all_products())
```

```python
# A new instance has access to the shared data in the class as well
pc2 = ProductCatalog()

# Print already-loaded shared data
print("Initial catalog:")
print(pc2.all_products())
```

```python
#or you can access it directly from the class
#
print(ProductCatalog.all_products())
```
