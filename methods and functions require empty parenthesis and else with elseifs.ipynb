### no else

```python
value = 10

if value > 20:
    print("Greater than 20")
elif value > 15:
    print("Greater than 15")
elif value > 5:
    print("Greater than 5")
```

### else

```python
value = 3

if value > 20:
    print("Greater than 20")
elif value > 15:
    print("Greater than 15")
elif value > 5:
    print("Greater than 5")
else:
    print("5 or less")
```
