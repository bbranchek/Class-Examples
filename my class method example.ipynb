```python
class Dog: 
    species = "Canis familiaris" 
    @classmethod 
    def get_species(cls): 
        return cls.species
```

```python
print(Dog.get_species())
```

```python
class Dog2: 
    species = "Canis familiaris" 
    def get_species(self): 
        return self.species
print(Dog2.get_species())    
```

```python
print(Dog2.get_species())
```

```python
fido=Dog2()
print(fido.get_species())
```

```python
class Animal: 
    def __init__(self, name): 
        self.name = name 
    @classmethod 
    def from_sound(cls, sound): 
            if sound == "woof": 
                return Dog("Doggo") 
            elif sound == "meow": 
                return Cat("Kitty") 
            return cls("Unknown") 
```

```python
class Dog(Animal): 
    pass 
class Cat(Animal): 
    pass 
a = Animal.from_sound("woof") 
print(type(a), a.name)
```
