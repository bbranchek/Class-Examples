```python
# local variable
x = 5
print(x)
def local1():
    x = 6
    print(x)
local1()
print(x)


```

```python
#Global variable
x = 5
print(x)
def local1():
    global x
    x = 6
    print(x)
local1()
print(x)
```

```python
#Incorrect use of nonlocal
x = 5
print(x)
def local1():
    nonlocal x
    x = 6
    print(x)
local1()
print(x)
```

x is global by default but not in a defined outer module

local1() has no enclosing function ... not defined outer function

therefore nonlocal x has nowhere valid to bind  error

nonlocal modifies variables in the nearest enclosing function

Your original code had no enclosing function, so nonlocal could not bind

```python
#Correct usage of nonlocal
def outer():
    x = 5
    print(x)

    def local1():
        nonlocal x
        x = 6
        print(x)

    local1()
    print(x)

outer()
```

```python
print(x)
```
