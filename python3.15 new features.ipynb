### non-lazy import

```python
import time

# --- Non-Lazy Import ---
t0 = time.perf_counter()
import pandas as pd # Executes and loads immediately
t1 = time.perf_counter()

print(f"Import time: {t1 - t0:.4f} seconds")  # e.g., ~2.0010 seconds
print("Program continues...")

t2 = time.perf_counter()
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 35, 40, 45],
    "City": ["New York", "Paris", "London", "Tokyo", "Sydney"],
}

df = pd.DataFrame(data)

# Use the .head() method to display the first 3 rows
print(df.head(3))  # Triggers the actual load/execution here
t3 = time.perf_counter()

print(f"First-use load time: {t3 - t2:.4f} seconds")  # e.g., ~2.0005 seconds

```

### lazy import

```python
import time

# --- Lazy Import (Python 3.15) ---
t0 = time.perf_counter()
lazy import pandas as pd # Binds name instantly, loads nothing yet
t1 = time.perf_counter()

print(f"Import time: {t1 - t0:.4f} seconds")  # e.g., ~0.0001 seconds (instant)
print("Program continues... (heavy module still not loaded)")

# Later in the code, when the module is actually needed:
t2 = time.perf_counter()
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 35, 40, 45],
    "City": ["New York", "Paris", "London", "Tokyo", "Sydney"],
}

df = pd.DataFrame(data)

# Use the .head() method to display the first 3 rows
print(df.head(3))  # Triggers the actual load/execution here
t3 = time.perf_counter()

print(f"First-use load time: {t3 - t2:.4f} seconds")  # e.g., ~2.0005 seconds
```

### Built-in frozendict

```python
# Create an immutable dictionary
config = frozendict({"host": "localhost", "port": 5432})
print(config["host"])

# Attempting to modify it raises a TypeError
try:
    config["port"] = 8080
except TypeError as e:
    print(f"Error: {e}")
```

###  Unpacking in Comprehensions

```python
nested_lists = [[1, 2], [3, 4], [5, 6]]

# Flatten 1 level deep using * unpacking
flat_list = [ *sublist for sublist in nested_lists ]
print(flat_list)  # Output: [1, 2, 3, 4, 5, 6]

# Dictionary unpacking in dict comprehensions
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3}
merged = { **d for d in (dict1, dict2) }
print(merged)  # Output: {'a': 1, 'b': 2, 'c': 3}
```

### Python 3.5 assumes your .py source file is encoded in UTF-8, allowing you to write Unicode characters directly inside string literals without special escape sequences

```python
# Save and run this file as UTF-8
message = "Hello, 世界! 🚀"
print(message)
```

### If your editor saves the file in a different encoding (like cp1252 or latin-1), you can declare the source file encoding explicitly at the top of the script:

```python
# -*- coding: utf-8 -*-
message = "Café"
print(message)
```

### In Python 3.5, the .encode() method on a string (str) and the .decode() method on a byte object (bytes) default to UTF-8

```python
# Converting a Python string to UTF-8 bytes
text = "Python 3.5 ✨"
encoded_bytes = text.encode()  # Defaults to 'utf-8'
print(encoded_bytes)
# Output: b'Python 3.5 \xe2\x9c\xa8'

# Explicitly specifying 'utf-8' (same result)
explicit_bytes = text.encode("utf-8")
print(explicit_bytes == encoded_bytes)
# Output: True

# Converting UTF-8 bytes back to a Python string
decoded_text = encoded_bytes.decode()  # Defaults to 'utf-8'
print(decoded_text)
# Output: Python 3.5 ✨
```

### Because Python 3.5's open() function relied on the system locale for text files (which could break portability on Windows), you must pass encoding="utf-8" to safely read and write UTF-8 files

```python
# Writing a text file with explicit UTF-8 encoding in Python 3.5
with open("example.txt", "w", encoding="utf-8") as f:
    f.write("Line 1: Grüßen\n")
    f.write("Line 2: 🌍 UTF-8 works great!\n")

# Reading the text file back with explicit UTF-8 encoding
with open("example.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print(content)
```
